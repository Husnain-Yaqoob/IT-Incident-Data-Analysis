## Data Type Analysis

```python
incident_df.dtypes
```

The data types of each column were examined to understand the structure of the dataset.

It was observed that some columns such as `reassignment_count`, `reopen_count`, and `sys_mod_count` are stored as numerical values, while columns like `active`, `made_sla`, and `knowledge` are stored as boolean values.

Additionally, the presence of many object-type columns indicates that the dataset contains a large number of categorical features, which may need to be encoded before applying machine learning models.

## Missing Value Analysis 

```python
incident_df.isnull().sum().sort_values(ascending=False)
```

```python
(incident_df == "?").sum().sort_values(ascending=False)
```

After checking for missing values using `isnull()`, it was found that there were no null values in the dataset. However, further analysis revealed that missing values were represented using the `"?"` symbol instead of actual null values.

A check was performed to count the number of `"?"` values in each column. The results showed that several columns contain a very high number of missing values.

Columns such as `caused_by`, `vendor`, `cmdb_ci`, `rfc`, and `problem_id` contain missing values in almost all rows (more than 99%).

Some columns such as `assigned_to`, `assignment_group`, and `u_symptom` contain a moderate number of missing values and may require further handling.

On the other hand, important columns such as `priority`, `impact`, `urgency`, and `incident_state` do not contain missing values and can be reliably used for analysis and modeling.

This step highlights the importance of checking for hidden missing values in real-world datasets.

## Understanding Incident Structure

```python
incident_df["number"].value_counts().head(10)
```

```python
incident_df["number"].nunique()
```

To understand the structure of the dataset, the number of occurrences of each incident was checked.

It was found that the same incident number appears multiple times. For example, some incidents appear more than 50 times.

The dataset contains 24,918 unique incidents, while the total number of rows is much higher.

This shows that the dataset is not a simple one-row-per-incident dataset, but an event log, where each row represents an update or change in an incident.

This is important because it affects how the data will be prepared for modeling.

This means that a decision will be needed later on whether to use event-level data or convert it into one row per incident for modeling.

## Duplicate Row Check
